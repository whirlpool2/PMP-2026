# Subiectul 2. Rețele Markov

## a)

$$P(A,B,C,D,E) = \frac{1}{Z}\,\phi_1(A,B)\,\phi_2(B,C)\,\phi_3(C,D)\,\phi_4(D,E)$$

Rețeaua Markov are câte o muchie între variabilele care apar împreună într-un potențial: $A - B$, $B - C$, $C - D$, $D - E$.

```python
# Lab05/Tema04/Ex2
import networkx as nx
import matplotlib.pyplot as plt

muchii = [("A", "B"), ("B", "C"), ("C", "D"), ("D", "E")]  # cate una pentru fiecare potential
graf = nx.Graph(muchii)

pozitii = {v: (i, 0) for i, v in enumerate("ABCDE")}  # pe o linie
plt.figure(figsize=(6, 1.5))
nx.draw(graf, pozitii, with_labels=True, node_color="#cfe3ff", node_size=1000, edge_color="#555")
plt.show()
```

Pornim de la definiția probabilității condiționate:

$$P(A \mid B, C) = \frac{P(A, B, C)}{P(B, C)}.$$

## b)

$$P(A,B,C,D,E) = \frac{1}{Z}\,\psi_1(A,B,C)\,\psi_2(B,D)\,\psi_3(C,E)$$

Variabilele din același potențial sunt vecine. $\psi_1$ are trei variabile, deci $A$, $B$, $C$ sunt conectate două câte două (triunghi). Din $\psi_2$ rezultă muchia $B - D$, iar din $\psi_3$ muchia $C - E$.

```python
# Lab05/Tema04/Ex2
import itertools
import networkx as nx
import matplotlib.pyplot as plt

factori = [("A", "B", "C"), ("B", "D"), ("C", "E")]  # variabilele fiecarui potential
muchii = [pereche for f in factori for pereche in itertools.combinations(f, 2)]
graf = nx.Graph(muchii)

pozitii = {"A": (0, 0), "B": (1, 1), "C": (1, -1), "D": (2, 1), "E": (2, -1)}
plt.figure(figsize=(4, 3))
nx.draw(graf, pozitii, with_labels=True, node_color="#cfe3ff", node_size=1000, edge_color="#555")
plt.show()
```

Pornim de la definiția probabilității condiționate:

$$P(D \mid A, B, C, E) = \frac{P(A, B, C, D, E)}{P(A, B, C, E)}.$$
