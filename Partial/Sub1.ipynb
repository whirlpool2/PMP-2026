## a)

```python
import numpy as np
from hmmlearn import hmm

stari = ["1", "2", "3", "4"] # camerele
culori = ["red", "green", "blue"] # observatiile

model = hmm.CategoricalHMM(n_components=4, init_params="")

# pozitia initiala este uniforma
model.startprob_ = np.array([1/4, 1/4, 1/4, 1/4])

# Matricea de tranzitie A[i, j] = P(camera_urmatoare=j | camera_curenta=i)
# Camerele sunt circulare: 1-2-3-4-1
model.transmat_ = np.array([
    [0.2, 0.4, 0.0, 0.4],
    [0.4, 0.2, 0.4, 0.0],
    [0.0, 0.4, 0.2, 0.4],
    [0.4, 0.0, 0.4, 0.2],
])

# B[i, k] = P(culoare=k | camera=i), ordine red, green, blue
model.emissionprob_ = np.array([
    [0.7, 0.2, 0.1],   # camera 1
    [0.2, 0.6, 0.2],   # camera 2
    [0.1, 0.3, 0.6],   # camera 3
    [0.5, 0.2, 0.3],   # camera 4
])
```

```python
secventa = ["red", "green", "blue", "red", "red"]

X = np.array([[culori.index(o)] for o in secventa])
# print("observatii codate:", X)

logprob, seq = model.decode(X, algorithm="viterbi")
camere = [stari[i] for i in seq]

for n, d in zip(secventa, camere):
    print(f"{n:>3}  ->  {d}")
print()
print("log-probabilitate:", logprob)
print("probabilitate:    ", np.exp(logprob))
```

## b)

Robotul nu mai poate rămâne pe loc, atunci matricea de tranziții are $0$ pe diagonala principală.

Deplasările posibile au probabilități egale, $0.5$.

```python
model2 = hmm.CategoricalHMM(n_components=4, init_params="")

# pozitia initiala este uniforma
model2.startprob_ = np.array([1/4, 1/4, 1/4, 1/4])

# Matricea de tranzitie A[i, j] = P(camera_urmatoare=j | camera_curenta=i)
# Camerele sunt circulare: 1-2-3-4-1
model2.transmat_ = np.array([
    [0.0, 0.5, 0.0, 0.5],
    [0.5, 0.0, 0.5, 0.0],
    [0.0, 0.5, 0.0, 0.5],
    [0.5, 0.0, 0.5, 0.0],
])

# B[i, k] = P(culoare=k | camera=i), ordine red, green, blue
model2.emissionprob_ = np.array([
    [0.7, 0.2, 0.1], #   1
    [0.2, 0.6, 0.2], #   2
    [0.1, 0.3, 0.6], #   3
    [0.5, 0.2, 0.3], #   4
])

secventa = ["red", "green", "blue", "red", "red"]

X = np.array([[culori.index(o)] for o in secventa])
# print("observatii codate:", X)

logprob2, seq2 = model2.decode(X, algorithm="viterbi")
camere2 = [stari[i] for i in seq2]

for n, d in zip(secventa, camere2):
    print(f"{n:>3}  ->  {d}")
print()
print("log-probabilitate:", logprob2)
print("probabilitate:    ", np.exp(logprob2))
```
