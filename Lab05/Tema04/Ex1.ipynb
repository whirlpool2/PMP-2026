## De la independență condiționată la factorizare

Într-o rețea Markov (graf neorientat), lipsa unei muchii directe între $A$ și $C$ reprezintă exact independența condiționată $A \perp C \mid B$: $A$ și $C$ comunică doar prin $B$.

## a) Ce înseamnă $A \perp C \mid B$?

Odată ce cunoaștem $B$, variabilele $A$ și $C$ devin independente:

$$P(a \mid b, c) = P(a \mid b) \qquad \text{(echivalent: } P(a, c \mid b) = P(a \mid b)\,P(c \mid b)\text{)}.$$

Dacă știm $B$, atunci $C$ nu aduce nicio informație nouă despre $A$: în lanțul $A - B - C$, $B$ este singura cale între ele.

## b) $A \perp C \mid B \;\Rightarrow\;$ factorizare

$$P(a, b, c) = P(b)\,P(a \mid b)\,P(c \mid a, b).$$

Folosim $A \perp C \mid B$. O formă echivalentă este $P(c \mid a, b) = P(c \mid b)$ ($a$ nu influențează pe $c$ dacă știm $b$).

$$P(a, b, c) = P(b)\,P(a \mid b)\,P(c \mid b).$$

$$\phi_{AB}(a, b) = P(a \mid b)\,P(b), \qquad \phi_{BC}(b, c) = P(c \mid b), \qquad Z = 1.$$

Atunci,

$$\phi_{AB}(a,b)\,\phi_{BC}(b,c) = P(a\mid b)\,P(b)\,P(c\mid b) = P(a,b,c) = \frac{1}{Z}\,\phi_{AB}(a,b)\,\phi_{BC}(b,c).$$

Fiindcă $P(a,b,c) > 0$ pentru orice configurație, rezultă $P(b) > 0$ și $P(a\mid b), P(c\mid b) > 0$. Deci $\phi_{AB}$ și $\phi_{BC}$ sunt strict pozitive, iar $Z = 1 > 0$.

## c) Reciproca: factorizare $\;\Rightarrow\; A \perp C \mid B$

Presupunem acum factorizarea și arătăm independența:

$$P(a, b, c) = \frac{1}{Z}\,\phi_{AB}(a, b)\,\phi_{BC}(b, c), \qquad \phi > 0.$$

Calculăm $P(a, c \mid b)$ și verificăm că e produsul $P(a\mid b)\,P(c\mid b)$.

Marginalizăm $a$ și $c$ (sumele se separă, fiecare potențial depinde de o singură variabilă liberă):

$$P(b) = \sum_{a, c} \frac{1}{Z}\,\phi_{AB}(a,b)\,\phi_{BC}(b,c) = \frac{1}{Z}\Big(\sum_{a}\phi_{AB}(a,b)\Big)\Big(\sum_{c}\phi_{BC}(b,c)\Big).$$

Notăm $\alpha(b) = \sum_{a}\phi_{AB}(a,b)$ și $\gamma(b) = \sum_{c}\phi_{BC}(b,c)$, deci $P(b) = \frac{1}{Z}\,\alpha(b)\,\gamma(b) > 0$.

$$P(a, c \mid b) = \frac{P(a,b,c)}{P(b)} = \frac{\frac{1}{Z}\,\phi_{AB}(a,b)\,\phi_{BC}(b,c)}{\frac{1}{Z}\,\alpha(b)\,\gamma(b)} = \frac{\phi_{AB}(a,b)}{\alpha(b)} \cdot \frac{\phi_{BC}(b,c)}{\gamma(b)}.$$

Sumând, recunoaștem cele două marginale:

$$P(a \mid b) = \sum_{c} P(a, c \mid b) = \frac{\phi_{AB}(a,b)}{\alpha(b)}, \qquad P(c \mid b) = \sum_{a} P(a, c \mid b) = \frac{\phi_{BC}(b,c)}{\gamma(b)}.$$

Deci $P(a, c \mid b) = P(a \mid b)\,P(c \mid b) \;\Longrightarrow\; A \perp C \mid B.$

## Concluzie

Pentru distribuții strict pozitive pe lanțul $A - B - C$:

$$A \perp C \mid B \quad\Longleftrightarrow\quad P(a,b,c) = \frac{1}{Z}\,\phi_{AB}(a,b)\,\phi_{BC}(b,c).$$

Structura grafului (ce muchii lipsesc) este echivalentă cu structura factorizării (pe ce grupuri de variabile se definesc potențialele).
