## a) Ce distribuție urmează $N$?

$N$ este numărul de pași până la (și incluzând) prima stemă. Fiecare aruncare este independentă și are aceeași probabilitate $p$ de a pica stemă, iar jocul se oprește la primul succes. Deci, $N$ urmează o distribuție geometrică cu probabilitatea $p$.

$$P(N = n) = (1 - p)^{n-1}\, p$$

## b) Simularea unui joc

```python
import random
import matplotlib.pyplot as plt
import numpy as np

def joaca_joc(p=0.5):
    N = 0
    S = 0.0
    while True:
        N += 1
        if random.random() < p:  # stemă
            z = random.randint(1, 6)
            S += z - 3
            break
        else:                     # ban
            S -= 0.5
    return N, S
```

```python
joaca_joc()
```

## c) Media lui $S$ și distribuția sa

Simulăm un număr mare de jocuri, estimăm media lui $S$.

```python
nr_jocuri = 100000
S_valori = [joaca_joc(p=0.5)[1] for _ in range(nr_jocuri)]

print(f"Media lui S: {np.mean(S_valori):.4f}")

plt.hist(S_valori, bins=40, color="lightgreen", edgecolor="black")
plt.title("Distribuția lui S (p = 0.5)")
plt.xlabel("S ($)")
plt.ylabel("Număr de jocuri")
plt.show()
```

## d) Monedă măsluită ($p = 0.3$, $p = 0.7$)

```python
for p in [0.3, 0.7]:
    S_valori = [joaca_joc(p=p)[1] for _ in range(nr_jocuri)]
    print(f"p = {p}  ->  media lui S: {np.mean(S_valori):.4f}")

    plt.hist(S_valori, bins=40, color="lightblue", edgecolor="black")
    plt.title(f"Distribuția lui S (p = {p})")
    plt.xlabel("S ($)")
    plt.ylabel("Număr de jocuri")
    plt.show()
```
