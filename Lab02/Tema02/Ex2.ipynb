# Tema 2 — Exercițiul 2

Un sistem de e-mail primește 30% mesaje spam. Cuvântul „ofertă” apare în 80% dintre mesajele spam și în 15% dintre mesajele legitime.

Notăm $S$ = mesaj spam, $L$ = mesaj legitim, cu $P(S) = 0.3$ și $P(L) = 0.7$.

$$P(\text{ofertă} \mid S) = 0.80, \qquad P(\text{ofertă} \mid L) = 0.15$$

## 1) $P(S \mid \text{ofertă})$

Aplicăm formula lui Bayes, cu numitorul dat de probabilitatea totală:

$$P(\text{ofertă}) = P(\text{ofertă} \mid S)\,P(S) + P(\text{ofertă} \mid L)\,P(L) = 0.80 \cdot 0.3 + 0.15 \cdot 0.7 = 0.345$$

$$P(S \mid \text{ofertă}) = \frac{P(\text{ofertă} \mid S)\,P(S)}{P(\text{ofertă})} = \frac{0.80 \cdot 0.3}{0.345} = \frac{0.24}{0.345} = \frac{16}{23} \approx 0.6957$$

Un mesaj care conține cuvântul „ofertă” este spam cu probabilitatea de aproximativ $69.6\%$.

## 2) $P(S \mid \text{ofertă}, \text{gratuit})$

Adăugăm $P(\text{gratuit} \mid S) = 0.60$ și $P(\text{gratuit} \mid L) = 0.10$. Presupunând **independența condiționată** a celor două cuvinte dat tipul mesajului:

$$P(\text{ofertă}, \text{gratuit} \mid S) = 0.80 \cdot 0.60 = 0.48, \qquad P(\text{ofertă}, \text{gratuit} \mid L) = 0.15 \cdot 0.10 = 0.015$$

$$P(\text{ofertă}, \text{gratuit}) = 0.48 \cdot 0.3 + 0.015 \cdot 0.7 = 0.1545$$

$$P(S \mid \text{ofertă}, \text{gratuit}) = \frac{0.48 \cdot 0.3}{0.1545} = \frac{0.144}{0.1545} = \frac{96}{103} \approx 0.9320$$

Prezența ambelor cuvinte crește probabilitatea de spam la aproximativ $93.2\%$.

## 3) De ce ipoteza independenței condiționate poate fi nerealistă

Independența condiționată presupune că apariția unui cuvânt nu oferă informații despre apariție celuilalt. Practic, cuvintele sunt corelate, pot apărea frecvent împreună ("ofertă gratuită"), deci există cazuri unde nu sunt independente.

## 4) Regulă de clasificare bazată pe probabilitatea condiționată

Clasificăm un mesaj drept spam dacă probabilitatea sa depășește un prag $\alpha$:

$$\text{spam} \iff P(S \mid \text{cuvinte}) > \alpha$$

Pragul poate fi ajustat în funcție de costul erorilor: un $\alpha$ mai mare reduce numărul de fals pozitive, iar un $\alpha$ mai mic prinde mai mult spam, dar cu risc mai mare de fals pozitive.
