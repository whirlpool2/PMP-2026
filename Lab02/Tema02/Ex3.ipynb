# Tema 2 — Exercițiul 3

O universitate analizează rata de promovare la două discipline, $A$ și $B$, pentru studenții din două programe de studiu:

| Program | Disciplina A | Disciplina B |
|---|---|---|
| Programul 1 | 81 din 90 | 19 din 20 |
| Programul 2 | 1 din 10 | 16 din 80 |

## 1) Rata de promovare pe disciplină, separat pe program

$$\text{Programul 1:} \quad A = \frac{81}{90} = 0.900, \qquad B = \frac{19}{20} = 0.950$$

$$\text{Programul 2:} \quad A = \frac{1}{10} = 0.100, \qquad B = \frac{16}{80} = 0.200$$

## 2) Care disciplină are rata mai mare în fiecare program

În ambele programe disciplina $B$ are rata de promovare mai mare:

$$\text{Programul 1:} \quad 0.950 > 0.900, \qquad \text{Programul 2:} \quad 0.200 > 0.100$$

## 3) Rata globală de promovare pe disciplină

$$A = \frac{81 + 1}{90 + 10} = \frac{82}{100} = 0.820, \qquad B = \frac{19 + 16}{20 + 80} = \frac{35}{100} = 0.350$$

La nivel global disciplina $A$ are rata mai mare ($0.820 > 0.350$). Ordinea se inversează față de comparația per program.

## 4) Inversarea ordinii disciplinelor

Deși disciplina $B$ are rata mai mare în **fiecare** program, la nivel global disciplina $A$ o depășește. Această inversare apare din cauza distribuției diferite a studenților:

- Disciplina $A$ are majoritatea studenților în Programul 1 (90 din 100), care are rate mari.
- Disciplina $B$ are majoritatea studenților în Programul 2 (80 din 100), care are rate mici.

Astfel rata globală a lui $A$ este trasă în sus de programul ușor, iar cea a lui $B$ este trasă în jos de programul dificil.

## 5) Formulare în termeni de probabilități condiționate

Fie $Pr$ promovarea, $D$ disciplina și $G$ programul. Comparația pe fiecare program folosește probabilitatea condiționată și pe program:

$$P(Pr \mid D = A, G) \quad \text{vs} \quad P(Pr \mid D = B, G)$$

iar comparația globală folosește probabilitatea marginală peste programe:

$$P(Pr \mid D) = \sum_{G} P(Pr \mid D, G)\, P(G \mid D)$$

Paradoxul lui Simpson apare pentru că ponderile $P(G \mid D)$ diferă între discipline: ordinea lui $P(Pr \mid D, G)$ pe fiecare program nu se păstrează neapărat în $P(Pr \mid D)$.

## Extensie Python: exemplu propriu de paradox Simpson

Construim un alt set de date care prezintă paradoxul: disciplina $A$ are rata mai mare în **fiecare** program, dar la nivel global disciplina $B$ este mai bună. Analizăm datele cu `pandas`.

```python
import pandas as pd

date = pd.DataFrame({
    "program":    ["Programul 1", "Programul 1", "Programul 2", "Programul 2"],
    "disciplina": ["A", "B", "A", "B"],
    "promovati":  [81, 234, 192, 55],
    "total":      [87, 270, 263, 80],
})
date["rata"] = date["promovati"] / date["total"]
date
```

```python
# Rata pe fiecare program (A vs B)
rate_program = date.pivot(index="program", columns="disciplina", values="rata")
rate_program
```

```python
# Rata globala, agregata peste programe
global_ = date.groupby("disciplina").agg(
    promovati=("promovati", "sum"),
    total=("total", "sum"),
)
global_["rata"] = global_["promovati"] / global_["total"]
global_
```

Pe fiecare program, disciplina $A$ are rata mai mare ($0.931 > 0.867$ și $0.730 > 0.688$), dar global disciplina $B$ o depășește ($0.826 > 0.780$).
