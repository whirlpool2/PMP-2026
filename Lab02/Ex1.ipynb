## a) Probabilitatea ca persoana să fie bolnavă după un test pozitiv

$$
\begin{aligned}
P(\text{Test} = \text{Pozitiv} \mid B) = 0.95 &\Longrightarrow P(\text{Test} = \text{Negativ} \mid B) = 0.05 \\
P(\text{Test} = \text{Negativ} \mid \neg B) = 0.90 &\Longrightarrow P(\text{Test} = \text{Pozitiv} \mid \neg B) = 0.10 \\
\end{aligned}
$$

Probabilitatea ca o persoană să fie bolnavă dacă este testată pozitivă $P(B \mid \text{Test} = \text{Pozitiv})$.

Formula lui Bayes: $P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$.

$$
P(B \mid \text{Test} = \text{Pozitiv}) = \frac{P(\text{Test} = \text{Pozitiv} \mid B) \cdot P(B)}{P(\text{Test} = \text{Pozitiv})}
$$

Probabilitatea ca un test să fie pozitiv este, conform legii probabilității totale, 

$$
\begin{aligned}
&P(\text{Test} = \text{Pozitiv}) = \\
=\ &P(\text{Test} = \text{Pozitiv} \mid B) \cdot P(B) + P(\text{Test} = \text{Pozitiv} \mid \neg B) \cdot P(\neg B) = \\
=\ &0.95 \cdot 0.01 + 0.10 \cdot 0.99 = \\
=\ &0.0095 + 0.099 = \\
=\ &0.1085
\end{aligned}
$$

Atunci,

$$
\begin{aligned}
&P(B \mid \text{Test} = \text{Pozitiv}) = \\
=\ &\frac{P(\text{Test} = \text{Pozitiv} \mid B) \cdot P(B)}{P(\text{Test} = \text{Pozitiv})} = \\
=\ &\frac{0.95 \cdot 0.01}{0.1085} = \\
\approx\ &0.087557
\end{aligned}
$$

Șansa ca o persoană care a testat pozitiv să fie bolnavă efectiv este de aproximativ $8.7\%$.

Deoarece specificitatea este de 90%, asta înseamnă că, din cei 99% oameni care sunt sănătoși, 10% sunt testați fals pozitiv. Pentru că proporția oamenilor sănătoși este mult mai mare, cei 10% testați fals pozitiv sunt mult mai mulți decât bolnavii testați 95% pozitiv.

## b) Modelați în Python probabilitatea de mai sus

```python
import numpy as np

def Not(x):
    return 1 - x

p_b = 0.01      # Șansa ca un om să fie bolnav
p_nb = Not(p_b) # Șansa ca un om să fie sănătos
p_poz_b = 0.95 # Sensibilitate
p_neg_nb = 0.9 # Specificitate

p_neg_b = Not(p_poz_b)   # Fals negativ
p_poz_nb = Not(p_neg_nb) # Fals pozitiv

p_poz = (p_poz_b * p_b) + (p_poz_nb * p_nb) # Șansa unui test pozitiv
p_b_poz = p_poz_b * p_b / p_poz # Bayes, șansa unui om bolnav dat un test pozitiv

print(f"P(B | Test = Pozitiv): {p_b_poz:.4%}")
```

Continuăm cu modelarea problemei:

```python
# Simulare

n = 1000000

# 1 = bolnav, 0 = sănătos
populatie = np.random.choice([1, 0], n, p = [p_b, p_nb])

bolnavi = populatie[populatie == 1]
sanatosi = populatie[populatie == 0]

# 1 = pozitiv, 0 = negativ
teste_bolnavi = np.random.choice([1, 0], len(bolnavi), p = [p_poz_b, p_neg_b])
teste_sanatosi = np.random.choice([1, 0], len(sanatosi), p = [p_poz_nb, p_neg_nb])

# P(B | Test = Pozitiv) = p_poz_b
tp = np.sum(teste_bolnavi == 1) 
fp = np.sum(teste_sanatosi == 1)

p_b_poz_simulat = tp / (tp + fp)
print(f"b) Simulat : P(B | Test = Pozitiv) = {p_b_poz_simulat:.4%}")
```

## c) Specificitatea minimă pentru P(B | Pozitiv) = 50%

```python
# target_p = P(B | Pozitiv)

target_p = 0.5

# Din formula lui Bayes: 
# P(B | Pozitiv) = (Sensibilitate * P(Bolnav)) / P(Test_Pozitiv_Total)
# target_p = (p_poz_b * p_b) / (p_poz_b * p_b + fals_pozitiv_max * p_nb)
# Izolăm fals_pozitiv_max:
fals_pozitiv_max = ((p_poz_b * p_b) / target_p - (p_poz_b * p_b)) / p_nb

# Specificitatea este 1 - rata de fals pozitiv
specificitate_minima = 1 - fals_pozitiv_max

print(f"c) Specificitatea minimă necesară: {specificitate_minima:.4%}")
```
