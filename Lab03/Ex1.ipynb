Rețeaua este:

$$S \rightarrow O, \qquad S \rightarrow L, \qquad S \rightarrow M, \qquad L \rightarrow M$$

$$P(S, O, L, M) = P(S)\,P(O \mid S)\,P(L \mid S)\,P(M \mid S, L)$$

![image.png](network.png)

## a) Independențe

Din structura grafului:

- **$O \perp L \mid S$**: $O$ și $L$ au cauza $S$
- **$O \perp M \mid S$**: singura legătură a lui $O$ cu restul rețelei trece prin $S$, $O \perp \{L, M\} \mid S$.

Variabilele $O$, $L$, $M$ nu sunt independente, fiindcă au cauza comună $S$.

## b) Clasificarea e-mail-urilor în funcție de $O$, $L$, $M$

Forma generală a formulii lui Bayes este:

$$
P(A \mid B_1, B_2, \dots, B_n)
= \frac{P(B_1, B_2, \dots, B_n \mid A)\, P(A)}{P(B_1, B_2, \dots, B_n)}
$$

iar numitorul se obține din legea probabilității totale:

$$
P(B_1, \dots, B_n) = \sum_{a} P(B_1, \dots, B_n \mid A=a)\, P(A=a)
$$

---

Probabilitatea $P(O, L, M \mid S)$ se descompune cu formula înmulțirii:

$$
P(O, L, M \mid S) = P(O \mid S)\, P(L \mid S, O)\, P(M \mid S, L, O)
$$

Dar...

- $P(L \mid S, O) = P(L \mid S)$, deoarece $O \perp L \mid S$
- $P(M \mid S, L, O) = P(M \mid S, L)$, deoarece $O \perp M \mid S, L$

Astfel probabilitatea devine:

$$
P(O, L, M \mid S) = P(O \mid S)\, P(L \mid S)\, P(M \mid S, L)
$$

Pentru un e-mail cu atributele observate $(O, L, M)$ calculăm probabilitatea folosind formula lui Bayes:

$$
P(S=1 \mid O, L, M)
= \frac{
    P(S=1)\, P(O \mid S=1)\, P(L \mid S=1)\, P(M \mid S=1, L)
}{
    \sum_{s \in \{0,1\}} P(S=s)\, P(O \mid S=s)\, P(L \mid S=s)\, P(M \mid S=s, L)
}
$$

Vom clasifica e-mailul ca fiind spam dacă $P(S=1 \mid O, L, M) > 0.5$.

De exemplu, pentru $(O=1, L=1, M=1)$:

$$
\begin{aligned}
P(S=1, O=1, L=1, M=1) &= 0.4 \cdot 0.7 \cdot 0.8 \cdot 0.9 = 0.2016 \\
P(S=0, O=1, L=1, M=1) &= 0.6 \cdot 0.1 \cdot 0.3 \cdot 0.6 = 0.0108 \\
P(S=1 \mid 1,1,1) &= \frac{0.2016}{0.2016 + 0.0108} = \frac{0.2016}{0.2124} \approx 0.9492 \;\Rightarrow\; \textbf{spam}
\end{aligned}
$$

Aplicând același calcul pentru toate cele 8 combinații $(O, L, M)$:


```python
P_S = {1: 0.4, 0: 0.6}
P_O = {1: 0.7, 0: 0.1}          # P(O=1 | S)
P_L = {1: 0.8, 0: 0.3}          # P(L=1 | S)
P_M = {(1, 1): 0.9, (1, 0): 0.5, (0, 1): 0.6, (0, 0): 0.2}  # P(M=1 | S, L)

# P + !P = 1
def prob(val, p1):
    return p1 if val == 1 else 1 - p1

for O in (0, 1):
    for L in (0, 1):
        for M in (0, 1):
            P_S_O_L_M = {} # P(S = s, O, L, M)
            for s in (0, 1):
                P_S_O_L_M[s] = (P_S[s]
                            * prob(O, P_O[s])
                            * prob(L, P_L[s])
                            * prob(M, P_M[(s, L)]))
            post = P_S_O_L_M[1] / (P_S_O_L_M[0] + P_S_O_L_M[1])
            eticheta = "spam" if post > 0.5 else "non-spam"
            print("O =", O, "L =", L, "M =", M, "=> P(S=1|O,L,M) =", round(post, 4), "=>", eticheta)
```
