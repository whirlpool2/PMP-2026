# Tema 2 — Exercițiul 1

Se aruncă două zaruri corecte, cu rezultatele $X$ și $Y$. Există $6 \times 6 = 36$ rezultate egal probabile.

## 1) $P(X + Y \geq 10 \mid X \text{ par})$

$X \in \{2, 4, 6\}$ (18 rezultate), numărăm câte au suma cel puțin 10:

$$P(X + Y \geq 10 \mid X \text{ par}) = \frac{4}{18} = \frac{2}{9} \approx 0.2222$$

## 2) $P(X + Y \geq 10 \mid \text{cel puțin un zar par})$

Evenimentul „cel puțin un zar par” este complementul lui „ambele impare” ($3 \times 3 = 9$ rezultate), deci are $36 - 9 = 27$ rezultate. Dintre acestea, 5 au suma cel puțin 10:

$$P(X + Y \geq 10 \mid \text{cel puțin un par}) = \frac{5}{27} \approx 0.1852$$

## 3) $P(\text{ambele pare} \mid X + Y = 8)$

Rezultatele cu suma 8 sunt $(2,6), (3,5), (4,4), (5,3), (6,2)$. Dintre ele, ambele numere pare: $(2,6), (4,4), (6,2)$.

$$P(\text{ambele pare} \mid X + Y = 8) = \frac{3}{5} = 0.6$$

## 4) Sunt $A = \{X + Y \geq 10\}$ și $B = \{X \text{ par}\}$ independente?

$A$ și $B$ sunt independente dacă $P(A \cap B) = P(A) \cdot P(B)$.

$$P(A) = \frac{6}{36} = \frac{1}{6}, \quad P(B) = \frac{18}{36} = \frac{1}{2}, \quad P(A) \cdot P(B) = \frac{1}{12}$$

$$P(A \cap B) = \frac{4}{36} = \frac{1}{9} \neq \frac{1}{12}$$

Evenimentele **nu sunt independente**.

## Extensie Python: simulare

Simulăm 100.000 de aruncări și estimăm probabilitățile condiționate, apoi le comparăm cu valorile exacte.

```python
import numpy as np

n = 100000
X = np.random.randint(1, 7, n)
Y = np.random.randint(1, 7, n)

suma = X + Y
cond1 = X % 2 == 0
cond2 = (X % 2 == 0) | (Y % 2 == 0)
cond3 = suma == 8

est1 = np.mean(suma[cond1] >= 10)
est2 = np.mean(suma[cond2] >= 10)
est3 = np.mean((X[cond3] % 2 == 0) & (Y[cond3] % 2 == 0))

print(f"1) simulat = {est1:.4f}   exact = {2/9:.4f}")
print(f"2) simulat = {est2:.4f}   exact = {5/27:.4f}")
print(f"3) simulat = {est3:.4f}   exact = {3/5:.4f}")
```
