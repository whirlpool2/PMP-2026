## Cum calculăm probabilitatea unei secvențe de observații? (Forward)

Model Markov ascuns cu stări ascunse $X_1, \dots, X_T$ și observații $O_1, \dots, O_T$. 

Notăm distribuția inițială $\pi_i = P(X_1 = i)$, probabilitățile de tranziție $a_{ij} = P(X_{t+1} = j \mid X_t = i)$ și probabilitățile de emisie $b_j(o) = P(O_t = o \mid X_t = j)$.

Pentru o secvență de observații $o_1, \dots, o_T$ definim variabila **Forward**

$$\alpha_t(j) = P(O_1 = o_1, \dots, O_t = o_t,\ X_t = j).$$

$\alpha_t(j)$ este probabilitatea comună a primelor $t$ observații și a faptului că la momentul $t$ ne aflăm în starea $j$.

## a) Inițializare: $\alpha_1(j) = \pi_j\, b_j(o_1)$

Pentru $t = 1$, definiția devine

$$\alpha_1(j) = P(O_1 = o_1,\ X_1 = j).$$

Aplicăm regula produsului (factorizarea unei probabilități comune):

$$\alpha_1(j) = P(X_1 = j)\, P(O_1 = o_1 \mid X_1 = j).$$

Primul factor este chiar distribuția inițială, $P(X_1 = j) = \pi_j$. Al doilea este probabilitatea de emisie din starea $j$, $P(O_1 = o_1 \mid X_1 = j) = b_j(o_1)$. Deci

$$\alpha_1(j) = \pi_j\, b_j(o_1).$$

## b) Recurență: $\alpha_t(j) = b_j(o_t) \sum_i \alpha_{t-1}(i)\, a_{ij}$

Notăm pe scurt $o_{1:t} = (o_1, \dots, o_t)$. Pornim de la definiție și introducem starea anterioară $X_{t-1}$ prin formula probabilității totale (adunăm peste stările $i$):

$$\alpha_t(j) = P(o_{1:t},\ X_t = j) = \sum_i P(o_{1:t},\ X_{t-1} = i,\ X_t = j).$$

Separăm, cu regula produsului, observația curentă $o_t$ și apoi tranziția $i \to j$:

$$P(o_{1:t},\ X_{t-1} = i,\ X_t = j) = \underbrace{P(o_{1:t-1},\ X_{t-1} = i)}_{\alpha_{t-1}(i)}\; \underbrace{P(X_t = j \mid \cdots)}_{a_{ij}}\; \underbrace{P(o_t \mid \cdots)}_{b_j(o_t)}.$$

Cei doi factori se simplifică prin proprietățile modelului:

- proprietatea Markov: $X_t$ depinde doar de $X_{t-1}$, deci $P(X_t = j \mid o_{1:t-1}, X_{t-1} = i) = a_{ij}$;
- proprietatea de emisie: $O_t$ depinde doar de $X_t$, deci $P(o_t \mid o_{1:t-1}, X_{t-1} = i, X_t = j) = b_j(o_t)$.

Înlocuind și scoțând $b_j(o_t)$ în fața sumei (nu depinde de $i$):

$$\alpha_t(j) = b_j(o_t) \sum_i \alpha_{t-1}(i)\, a_{ij}, \qquad t = 2, \dots, T.$$

## c) Finalizare: $P(o_1, \dots, o_T) = \sum_j \alpha_T(j)$

La momentul final $T$, $\alpha_T(j) = P(o_{1:T},\ X_T = j)$. Starea finală $X_T$ poate fi oricare, deci:

$$P(o_{1:T}) = \sum_j P(o_{1:T},\ X_T = j) = \sum_j \alpha_T(j).$$

Probabilitatea întregii secvențe observate se obține adunând valorile Forward de pe ultima coloană.

## Concluzie

Diferența față de Viterbi: Forward adună peste stările precedente (probabilitatea totală a observațiilor), pe când Viterbi maximizează probabilitatea unui singur traseu.
