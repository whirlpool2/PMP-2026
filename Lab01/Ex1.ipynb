### a) Simulați în Python experimentul.

Considerăm o urnă în care avem 3 bile roşii, 4 albastre şi 2 negre. Aruncăm un zar; dacă obţinem un număr prim, adăugăm o bilă neagră în urnă, dacă obţinem 6, adăugăm o bilă roşie, iar în celelalte cazuri adăugăm o bilă albastră.

Apoi extragem o bilă din urnă.

```python
import random

def simulate(iterations = 100000):
    red_count = 0
    
    for _ in range(iterations):
        urn = ["R"] * 3 + ["B"] * 4 + ["N"] * 2
        dice = random.randint(1, 6)
        
        if dice in (2, 3, 5):
            urn.append("N")
        elif dice == 6:
            urn.append("R")
        else:
            urn.append("B")

        ball = random.choice(urn)
        
        if ball == "R":
            red_count += 1
        
    return red_count / iterations
```

### b) Folosind modelarea făcută, estimaţi probabilitatea de a obţine o bilă roşie.

```python
estimated_p = simulate()
print(f"Probabilitatea estimată de a obține o bilă roșie este {estimated_p:.4f}")
```

### c) Calculaţi probabilitatea teoretică a evenimentului de mai sus şi comparaţi-o cu cea obţinută.

Pentru evenimentul $R$ (extragem bila roșie), avem trei cazuri la aruncarea zarului:
- Obținem prim ($2, 3, 5$), adăugăm neagră. $P(R \mid \text{prim}) = \frac{3}{10}$
- Obținem $6$, adăugăm roșie. $P(R \mid 6) = \frac{4}{10}$.
- Obținem $1, 4$, adăugăm albastră. $P(R \mid \{1,4\}) = \frac{3}{10}$.

Formula probabilității totale este:

$$
P(R) = \frac{1}{2}\cdot\frac{3}{10} + \frac{1}{6}\cdot\frac{4}{10} + \frac{1}{3}\cdot\frac{3}{10} = \frac{19}{60} \approx 0.3167
$$

```python
theoretical_p = (1/2) * (3/10) + (1/6) * (4/10) + (1/3) * (3/10)

print(f"Probabilitatea estimată: {estimated_p:.4f}")
print(f"Probabilitatea teoretică: {theoretical_p:.4f}")
print(f"Diferență: {theoretical_p - estimated_p:.4f}")
```
