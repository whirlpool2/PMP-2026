## Pregătirea datelor

Dacă nu avem acces la listă, generăm un fișier `studenti.csv` cu nume generice. Fișierul are două coloane: `nume` și `a_prezentat` (`da`/`nu`).

```python
import csv
import random

with open("studenti.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["nume", "a_prezentat"])
    for i in range(1, 16):
        writer.writerow([f"Student{i:02d}", random.choice(["da", "nu"])])
```

## Citirea fișierului și selecția aleatoare

Citim fișierul, păstrăm doar studenții care **nu** au prezentat încă, apoi extragem `k` nume fără repetiție cu `random.sample`.

```python
neprezentati = []
with open("studenti.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for rand in reader:
        if rand["a_prezentat"] == "nu":
            neprezentati.append(rand["nume"])

neprezentati
```

```python
k = 3
selectie = random.sample(neprezentati, k) # alege poziții distincte 
selectie
```
